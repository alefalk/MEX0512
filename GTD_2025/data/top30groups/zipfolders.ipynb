```python
import zipfile
import os

folder_to_zip = "anonLocations"
zip_filename = "anonLocations.zip"

with zipfile.ZipFile(zip_filename, 'w', zipfile.ZIP_DEFLATED) as zipf:
    for root, dirs, files in os.walk(folder_to_zip):
        for file in files:
            full_path = os.path.join(root, file)
            relative_path = os.path.relpath(full_path, folder_to_zip)
            zipf.write(full_path, arcname=os.path.join(folder_to_zip, relative_path))

print(f"Zipped folder to {zip_filename}")
```
